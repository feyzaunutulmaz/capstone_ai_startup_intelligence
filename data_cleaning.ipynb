```python
df_invest_raw = df_invest_raw.drop_duplicates()
```

```python
df_aistartup_raw['deal_date'] = pd.to_datetime(df_aistartup_raw['deal_date'])
```

```python
df_invest_raw.dtypes
```

```python
df_invest_raw.columns = df_invest_raw.columns.str.strip()
```

```python
df_invest_raw["founded_at"] = pd.to_datetime(df_invest_raw["founded_at"], errors="coerce")

df_invest_raw["first_funding_at"] = pd.to_datetime(df_invest_raw["first_funding_at"], errors="coerce")

df_invest_raw["last_funding_at"] = pd.to_datetime(df_invest_raw["last_funding_at"], errors="coerce")

df_invest_raw["funding_total_usd"] = pd.to_numeric(
    df_invest_raw["funding_total_usd"]
    .str.replace(",", "", regex=False),
    errors="coerce"
)
```

```python
df_invest_raw.dtypes
```

```python
df_invest_raw["founded_year"] = pd.to_numeric(df_invest_raw["founded_year"], errors="coerce").astype("Int64")
df_invest_raw["founded_month"] = pd.to_datetime(df_invest_raw["founded_month"], format="%Y-%m", errors="coerce")
```

```python
df_invest_raw.dtypes
```

```python
df_invest_raw[
    df_invest_raw["market"].str.contains("Artificial Intelligence|Machine Learning|Deep Learning|Generative AI", case=False, na=False) |
    df_invest_raw["category_list"].str.contains("Artificial Intelligence|Machine Learning|Deep Learning|Generative AI", case=False, na=False)
]
```

```python
print("Artificial Intelligence:", df_invest_raw["category_list"].str.contains("Artificial Intelligence", case=False, na=False).sum())

print("Machine Learning:", df_invest_raw["category_list"].str.contains("Machine Learning", case=False, na=False).sum())

print("Deep Learning:", df_invest_raw["category_list"].str.contains("Deep Learning", case=False, na=False).sum())

print("Generative AI:", df_invest_raw["category_list"].str.contains("Generative AI", case=False, na=False).sum())

print("Natural Language Processing:", df_invest_raw["category_list"].str.contains("Natural Language Processing", case=False, na=False).sum())

print("AI:", df_invest_raw["category_list"].str.contains(r"\bAI\b", case=False, na=False).sum())
```
