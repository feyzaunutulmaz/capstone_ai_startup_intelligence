We encountered column names with excess of spaces which are being removed now:

```python
df_invest_raw.columns = df_invest_raw.columns.str.strip()
```

The datatype of all dates have to be changed to datetime & funding_total_usd has to be changed to a float and also the syntax has to be corrected:

```python
df_invest_raw["founded_at"] = pd.to_datetime(df_invest_raw["founded_at"], errors="coerce")

df_invest_raw["first_funding_at"] = pd.to_datetime(df_invest_raw["first_funding_at"], errors="coerce")

df_invest_raw["last_funding_at"] = pd.to_datetime(df_invest_raw["last_funding_at"], errors="coerce")

df_invest_raw["funding_total_usd"] = pd.to_numeric(
    df_invest_raw["funding_total_usd"]
    .str.replace(",", "", regex=False),
    errors="coerce"
)
```

Add new columns:

Adding primary subsector column:

```python
df_invest_raw['category_list_clean'] = df_invest_raw['category_list_clean'].apply(
    lambda x: x if isinstance(x, list) else []
)
```

```python
generic = {'Artificial Intelligence', 'Machine Learning', 'Software', 'Technology'}
df_invest_raw['primary_subsector'] = df_invest_raw['category_list_clean'].apply(
    lambda tags: next((t for t in tags if t not in generic), tags[0] if tags else None)
)
df_invest_raw['n_categories'] = df_invest_raw['category_list_clean'].apply(len)
```

```python
df_invest_raw.primary_subsector
```

Adding funding stage column:

```python
round_cols = ['seed','angel','venture','round_A','round_B','round_C','round_D',
              'round_E','round_F','round_G','round_H','private_equity',
              'post_ipo_equity','post_ipo_debt','debt_financing',
              'convertible_note','grant','equity_crowdfunding',
              'product_crowdfunding','secondary_market','undisclosed']

# höchste je erreichte Runde (grobes Reife-Maß)
letter_rounds = ['round_A','round_B','round_C','round_D','round_E','round_F','round_G','round_H']
def furthest_round(row):
    reached = [c for c in letter_rounds if row[c] > 0]
    return reached[-1].replace('round_', '') if reached else ('Seed/Angel only' if row['seed']+row['angel']>0 else 'Other/Unknown')
df_invest_raw['furthest_round_reached'] = df_invest_raw.apply(furthest_round, axis=1)
```

```python
df_invest_raw.furthest_round_reached
```

Adding average funding per round column:

```python
df_invest_raw['avg_funding_per_round'] = df_invest_raw['funding_total_usd'] / df_invest_raw['funding_rounds']
```

```python
df_invest_raw.avg_funding_per_round
```

Adding threshold of highly funded companies:

```python
# Schwellenwert definieren, z. B. oberstes Quartil oder Top 10%
threshold = df_invest_raw['funding_total_usd'].quantile(0.75)
df_invest_raw['highly_funded'] = df_invest_raw['funding_total_usd'] >= threshold
```

```python
df_invest_raw.highly_funded.tail(5)
```

Adding column: Operating (1) or Aquired/Closed (0)

```python
df_invest_raw['is_operating'] = (df_invest_raw['status'] == 'operating').astype(int)
```

```python
df_invest_raw.is_operating.tail(5)
```

Adding column für company age at last funding and years to first funding:

```python
df_invest_raw['company_age_at_last_funding'] = (df_invest_raw['last_funding_at'] - df_invest_raw['founded_at']).dt.days / 365.25
df_invest_raw['years_to_first_funding'] = (df_invest_raw['first_funding_at'] - df_invest_raw['founded_at']).dt.days / 365.25
```

```python
df_invest_raw.company_age_at_last_funding
```

In case of negative numbers because of anamolies we change the negative outcomes to a NaN:

```python
df_invest_raw.loc[df_invest_raw['company_age_at_last_funding'] < 0, 'company_age_at_last_funding'] = pd.NA
df_invest_raw.loc[df_invest_raw['years_to_first_funding'] < 0, 'years_to_first_funding'] = pd.NA
```
