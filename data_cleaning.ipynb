Add new columns:

Adding primary subsector column:

```python
df_invest_raw['category_list'] = df_invest_raw['category_list'].apply(
    lambda x: x if isinstance(x, list) else []
)
```

```python
generic = {'Artificial Intelligence', 'Machine Learning', 'Software', 'Technology'}
df_invest_raw['primary_subsector'] = df_invest_raw['category_list'].apply(
    lambda tags: next((t for t in tags if t not in generic), tags[0] if tags else None)
)
df_invest_raw['n_categories'] = df_invest_raw['category_list'].apply(len)
```

```python
df_invest_raw.primary_subsector
```

Adding funding stage column:

```python
round_cols = ['seed','angel','venture','round_A','round_B','round_C','round_D',
              'round_E','round_F','round_G','round_H','private_equity',
              'post_ipo_equity','post_ipo_debt','debt_financing',
              'convertible_note','grant','equity_crowdfunding',
              'product_crowdfunding','secondary_market','undisclosed']

# höchste je erreichte Runde (grobes Reife-Maß)
letter_rounds = ['round_A','round_B','round_C','round_D','round_E','round_F','round_G','round_H']
def furthest_round(row):
    reached = [c for c in letter_rounds if row[c] > 0]
    return reached[-1].replace('round_', '') if reached else ('Seed/Angel only' if row['seed']+row['angel']>0 else 'Other/Unknown')
df_invest_raw['furthest_round_reached'] = df_invest_raw.apply(furthest_round, axis=1)
```

Exploratory Analysis

```python

import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import Image, display
# We display charts via IPython.display.Image on the saved PNG instead of
# plt.show(), since plt.show() depends on the notebook's matplotlib backend
# and can misbehave (showing a link/object repr instead of the plot). Reading
# back the saved file always works regardless of backend.
 
plt.rcParams.update({
    'font.size': 11,
    'axes.spines.top': False,
    'axes.spines.right': False,
    'axes.edgecolor': '#555555',
    'figure.facecolor': 'white',
})
BLUE = '#3b6ea5'
ORANGE = '#e08e45'
 
df = df_invest_raw.copy()
df['founded_year'] = pd.to_numeric(df['founded_year'], errors='coerce')
 
# 1. Top 10 AI sectors ("market") by total funding
top_market = df.groupby('market')['funding_total_usd'].sum().sort_values(ascending=False).head(10) / 1e6
fig, ax = plt.subplots(figsize=(7, 4.2))
top_market[::-1].plot(kind='barh', ax=ax, color=BLUE)
ax.set_xlabel('Total Funding (USD million)')
ax.set_ylabel('')
ax.set_title('Top 10 AI Sectors by Total Funding')
plt.tight_layout()
plt.savefig('chart_top_markets_en.png', dpi=180)
plt.close()
display(Image('chart_top_markets_en.png'))
```

```python
# 2. Startups founded & total funding by founding year
by_year = df.dropna(subset=['founded_year']).groupby('founded_year').agg(
    n=('name', 'count'), total_funding=('funding_total_usd', 'sum')
)
by_year = by_year[by_year.index >= 1995]
fig, ax1 = plt.subplots(figsize=(10, 4))
ax1.bar(by_year.index, by_year['n'], color=BLUE, alpha=0.8, label='Startups founded')
ax1.set_ylabel('Number of startups founded', color=BLUE)
ax1.set_xlabel('Founding year')
ax2 = ax1.twinx()
ax2.plot(by_year.index, by_year['total_funding'] / 1e6, color=ORANGE, marker='o',
         linewidth=2, label='Total funding')
ax2.set_ylabel('Total funding (USD million)', color=ORANGE)
ax1.set_title('Startups Founded & Funding by Founding Year')
plt.tight_layout()
plt.savefig('chart_funding_over_time_en.png', dpi=180)
plt.close()
display(Image('chart_funding_over_time_en.png'))
```

```python

# 3. Top 10 countries by number of AI startups
top_country = df[df['country_code'] != 'Unknown']['country_code'].value_counts().head(10)
fig, ax = plt.subplots(figsize=(10, 4))
top_country[::-1].plot(kind='barh', ax=ax, color=BLUE)
ax.set_xlabel('Number of startups')
ax.set_title('Top 10 Countries by Number of AI Startups')
plt.tight_layout()
plt.savefig('chart_top_countries_en.png', dpi=180)
plt.close()
display(Image('chart_top_countries_en.png'))
```

```python

# 4. Company status distribution (pie chart)
status_counts = df['status'].value_counts()
fig, ax = plt.subplots(figsize=(5.5, 4.2))
colors_list = [BLUE, ORANGE, '#8fbf8f', '#c0c0c0']
ax.pie(status_counts.values, labels=status_counts.index, autopct='%1.0f%%',
       colors=colors_list[:len(status_counts)], startangle=90, textprops={'fontsize': 10})
ax.set_title('Company Status Distribution')
plt.tight_layout()
plt.savefig('chart_status_en.png', dpi=180)
plt.close()
display(Image('chart_status_en.png'))
```

```python
# 5. Distribution of furthest funding round reached
round_order = ['Seed/Angel only', 'A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'Other/Unknown']
round_counts = df['furthest_round_reached'].value_counts().reindex(round_order).dropna()
fig, ax = plt.subplots(figsize=(7, 4))
round_counts.plot(kind='bar', ax=ax, color=BLUE)
ax.set_ylabel('Number of startups')
ax.set_xlabel('Furthest funding round reached')
ax.set_title('Distribution of Furthest Funding Round Reached')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('chart_rounds_en.png', dpi=180)
plt.close()
display(Image('chart_rounds_en.png'))
```

```python

# 6. Funding rounds: highly funded vs. rest
grp = df.groupby('highly_funded')['funding_rounds'].mean()
fig, ax = plt.subplots(figsize=(5.5, 4.2))
labels = ['Not highly funded\n(< Q3)', 'Highly funded\n(top 25%)']
ax.bar(labels, [grp.get(False, 0), grp.get(True, 0)], color=[BLUE, ORANGE])
ax.set_ylabel('Avg. number of funding rounds')
ax.set_title('Funding Rounds: Highly Funded vs. Rest')
plt.tight_layout()
plt.savefig('chart_highly_funded_en.png', dpi=180)
plt.close()
display(Image('chart_highly_funded_en.png'))
```

Create lead investor table (one row per investor per deal) for Tableau:


```python
df_lead_investors = df_aistartup_raw[['company', 'round_type', 'deal_size_tier',
                                      'funding_total_usd', 'lead_investors']].copy()

# split multiple investors into separate rows
df_lead_investors['lead_investor'] = df_lead_investors['lead_investors'].str.split(',')
df_lead_investors = df_lead_investors.explode('lead_investor')
df_lead_investors['lead_investor'] = df_lead_investors['lead_investor'].str.strip()

# harmonize spelling
df_lead_investors['lead_investor'] = df_lead_investors['lead_investor'].replace({
    'a16z': 'Andreessen Horowitz',
    'Lightspeed': 'Lightspeed Venture Partners',
    'Google (acquirer)': 'Google',
    'Meta (49% stake)': 'Meta'
})

# remove entries that are not actual investors
exclude = ['Public Market (NYSE)', 'Public Market (HKEX)', 'Undisclosed',
           'Various (undisclosed)', 'Private investor (undisclosed)']
df_lead_investors = df_lead_investors[~df_lead_investors['lead_investor'].isin(exclude)]

df_lead_investors['lead_investor'].value_counts().head(10)
```
